# MiniCOCO Dataset Overview

### COCO 2017 dataset

The COCO (Common Objects in Context) dataset is a large-scale image recognition dataset used for various computer vision tasks like object detection, segmentation, and captioning. It contains over 330,000 images, each annotated with 80 object categories and 5 captions describing the scene.

COCO is benchmark dataset commonly used in machine learning—both for research and practical applications.

Overview of COCO 2017 dataset is given in [this blogpost](https://www.v7labs.com/blog/coco-dataset-guide) and in [paper 'Microsoft COCO: Common Objects in Context'](https://arxiv.org/abs/1405.0312).

### MiniCOCO overview

Dataset used in this project is miniCOCO. It contains 100 train and 20 validation images and its captions.

Train and validation datasets in miniCOCO are **random subsets** of train and validation [COCO 2017 datasets](https://cocodataset.org/#home). The script for downloading subset of COCO images and their corresponding captions is provided in the file 'download_coco_subset.py'. **Don't run 'download_coco_subset.py' if you use provided dataset.** 

Each image has five captions stored in files 'train_captions.csv' and 'val_captions.csv'

### Display Images from Train Dataset with Captions

A single image has numerous different options for caption. In miniCOCO dataset 5 captions are assigned to each image manually.

We will pick a one image from miniCOCO and display it with its captions. 

```python
import pandas as pd
import matplotlib.pyplot as plt
import skimage.io as io
```

```python
csv_train_path = f'miniCOCO/train_captions.csv'
frame = pd.read_csv(csv_train_path)  # DataFrame contains all miniCOCO train images names and their corresponding captions
```

```python
len(frame)  # Number of images
```

```python
img_name = frame.iloc[0]['img_name']  # Name of the first image in miniCOCO train dataset 
img_name
```

```python
frame.iloc[0]['caption1'] # The first caption of the first image in miniCOCO train dataset 
```

```python
frame.iloc[0]['caption3'] # The third caption of the first image in miniCOCO train dataset 
```

```python
img_path = f'miniCOCO/train/{img_name}'
I = io.imread(img_path)
type(I), I.shape  # I is (w, h, 3) NumPy ndarray
```

```python
# Display the image
plt.figure(figsize=(5, 5))
plt.axis('off')
plt.imshow(I)
plt.figtext(0.5, 0.1, frame.iloc[0, 1:]['caption1'], ha='center', va='bottom', fontsize=7) # raise a little
plt.figtext(0.5, 0.075, frame.iloc[0, 1:]['caption2'], ha='center', va='bottom', fontsize=7) # raise a little less
plt.figtext(0.5, 0.05, frame.iloc[0, 1:]['caption3'], ha='center', va='bottom', fontsize=7)
plt.figtext(0.5, 0.025, frame.iloc[0, 1:]['caption4'], ha='center', va='bottom', fontsize=7)
plt.figtext(0.5, 0.0, frame.iloc[0, 1:]['caption5'], ha='center', va='bottom', fontsize=7)
plt.show()
```
